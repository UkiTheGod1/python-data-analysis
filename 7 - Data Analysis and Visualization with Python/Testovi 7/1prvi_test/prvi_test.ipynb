# 🥚1. Korak
Prvo i osnovno je da moramo **učitati podatke** iz našeg fajla "books.csv", naravno, pomoću pandas-a.

```python
import pandas as pd
df = pd.read_csv('books.csv')
```

# 🐣2. Korak
Sada ćemo grupisati redove po sekciji pomoću **".groupby"** i primenićemo određene funkcije na svaku grupu pomoću **".agg()".**

Za redove u koloni **"title"** ćemo samo izračunati koliko ih ima **(count)**, a za kolonu **"times_borrowed"** ćemo sabrati sve vrednosti **(sum)**.

Time dobijamo ukupan broj knjiga i ukupan broj iznajmljivanja po sekciji.

```python
inventory_gap = df.groupby('section').agg({
'title': 'count',
'times_borrowed': 'sum'
})
```

# 🐥3. korak
Kada smo sve grupisali kako treba, napravićemo novu kolonu pod nazivom **"titles_to_borrow_ratio"** koja će prikazivati odnos broja iznajmljivanja i broja knjiga u svakoj sekciji deljenjem te dve vrednosti.

Time dobijamo koliko puta je svaka knjiga pozajmljena iz određene sekcije u proseku.

```python
inventory_gap['titles_to_borrow_ratio'] = inventory_gap['times_borrowed'] / inventory_gap['title']
```

# 🐔4. Korak
Na kraju ćemo samo izvršiti ispis kako bismo pregledali naš kod. Sortiraćemo vrednosti na osnovu nove kolone i tako da najveće vrednosti budu na vrhu.

```python
print(inventory_gap.sort_values(by=['titles_to_borrow_ratio'], ascending=False))
```
