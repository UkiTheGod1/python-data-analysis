# 🥚 Prvi korak:
Moramo da importujemo bazu podataka u Python kako bismo mogli da vrsimo operacije nad njom:

```python
import pandas as pd
df = pd.read_csv('books.csv')
```

# 🐣 Drugi korak:
Sada cemo da napravimo tu novu kolona koja ce imati naziv "Borrowings per copy"

```python
df['borrowings_per_copy'] = df['times_borrowed'] / df['total_copies']
```

# 🐥 Treci korak:
Potrebno je prikazati novu kolonu na jedan pregladan nacin. Sortiracemo vrednosti na osnovnu nove kolone i ispisacemo prvih 50 redova sa najvecim rezultatom. Takodje cemo izabrati ispis samo odredjenih kolona.

```python
sorted_df = df.sort_values(by='borrowings_per_copy', ascending=False)
top_50 = sorted_df.head(50)
top_50[['title', 'times_borrowed', 'total_copies', 'borrowings_per_copy']]
```
