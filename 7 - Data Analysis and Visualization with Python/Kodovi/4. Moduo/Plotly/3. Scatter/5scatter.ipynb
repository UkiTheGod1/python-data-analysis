```python
import pandas as pd
import plotly.express as px
 
df = pd.read_csv("books.csv")
 
fig = px.scatter(df, x="page_count", y="rating", title="Rating vs Page Count")
fig.show()
```

## Hover
Dodaćemo "hover name" koji će prikazivati "title" boldovanim iznad svih podataka kao naziv tačke, i dodaćemo ostale podatke hovera (hover_data)

```python
import pandas as pd
import plotly.express as px
 
df = pd.read_csv("books.csv")
 
fig = px.scatter(df, x="page_count", y="rating", 
                 hover_name="title",
                 hover_data=["author", "genre", "ratings_count", "year_published", "page_count", "price", "language"], 
                 title="Rating vs Page Count")
fig.show()
```

## Trendline - Linija trenda
Linija trenda (trendline) pomaže nam da odmah uočimo:
- Da li ocene rastu sa brojem stranica?
- Ili možda obrnuto – što više strana, ocene padaju?
- Ili nema nikakve veze – sve je rasejano bez logike?

```python
import pandas as pd
import plotly.express as px
 
df = pd.read_csv("books.csv")
 
fig = px.scatter(df, x="page_count", y="rating", 
                 hover_name="title",
                 hover_data=["author", "genre", "ratings_count", "year_published", "page_count", "price", "language"], 
                 trendline="ols",
                 title="Rating vs Page Count")
fig.show()
```

## Više dijagrama u jednom prikazivanju
Koristimo plotly.graph_objects umesto plotly.express

```python
import pandas as pd
import plotly.graph_objects as go

df = pd.read_csv("books.csv")
 
fig = go.Figure()
features = ["page_count", "ratings_count", "year_published", "price"]
 
# hover data
hover_text = df.apply(lambda row:
                      f"<b>{row['title']}</b><br><br>"
                      f"Author: {row['author']}<br>"
                      f"Pages: {row['page_count']}<br>"
                      f"Price: {row['price']}<br>"
                      f"Year: {row['year_published']}<br>"
                      f"Ratings count: {row['ratings_count']}", axis=1)
 
# create trace (plot) for each feature
for i, feature in enumerate(features):
    fig.add_trace(go.Scatter(
        x=df[feature],
        y=df["rating"],
        mode="markers",
        hovertext=hover_text,
        hoverinfo="text",
        name=feature,
        visible=(i == 0)  # only first trace will be visible
    ))
 
  
# create buttons
buttons = []
for i, feature in enumerate(features):
    visibility = [j == i for j in range(len(features))]
    buttons.append(dict(
        label=feature,
        method="update",
        args=[
            {"visible": visibility},
            {
                "title": {"text": f"Rating vs {feature}"},
                "xaxis": {"title": {"text": feature}}
            }
        ]
    ))
 
# add menu and configure
fig.update_layout(
    updatemenus=[dict(
        type="dropdown",
        direction="down",
        buttons=buttons,
        x=1.15,
        y=0.9
    )],
    title=f"Rating vs {features[0]}",
    xaxis_title=features[0],
    yaxis_title="Rating",
    showlegend=False
)

fig.show()
```
