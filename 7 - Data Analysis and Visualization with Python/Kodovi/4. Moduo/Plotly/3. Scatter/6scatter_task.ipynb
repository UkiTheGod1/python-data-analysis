## Vezbamo Scatter sa Plotly-em
Setimo se skupa podataka o studentskim navikama i ocenama, koji smo koristili ranije da ispitamo kako različite navike utiču na uspeh. Tada smo koristili statičke grafikone. Sada je vreme za interaktivni upgrade!

student_performance_data.csv

Zadatak: Iskoristimo moć interaktivnih tačkastih dijagrama kako bismo otkrili:

- koji faktori najviše utiču na studentski uspeh;
- koje korelacije bi nam promakle bez dinamičkog istraživanja;
- i ko se krije iza tih najzanimljivijih tačkica.

Šta je potrebno da uradimo?

- Učitajmo skup podataka student_performance_data.csv.
- Kreirajmo interaktivne scatter plotove unutar jedne figure.
- Aktivirajmo hover prikaz sa detaljima (ime i prezime, univerzitet, godine, broj časova, vreme provedeno u učenju...).
- Dodajmo padajući meni za biranje različitih karakteristika koje poredimo sa ocenom (grade).

Kada postaviš dijagrame, pokušaj da odgovoriš na sledeća pitanja – koristi hover i pametno zumiranje:

- Ko su i koje univerzitete pohađaju tri studenta sa najboljim ocenama?
- Ko je proveo najviše vremena u učenju, a nije položio?
- Koju ocenu je ostvario student koji je propustio najviše časova?
- Ko je najbolji student s brojem obuće 46? (Da, postoji i ta kolona!)

```python
import plotly.graph_objects as go
import pandas as pd
df = pd.read_csv("student_performance_data.csv")

fig = go.Figure()
features = ["study_hours", "missed_classes", "sleep_hours", "shoe_size"]

hover_text= df.apply(lambda row:
                     f"<b>{row['full_name']}</b><br><br>"
 
                      f"University: {row['university']}<br>"
 
                      f"Age: {row['age']}<br>"
 
                      f"Study hours: {row['study_hours']:.2f}<br>"
 
                      f"Missed classes: {row['missed_classes']}<br>"
 
                      f"Sleep hours: {row['sleep_hours']:.2f}<br>"
 
                      f"Shoe size: {row['shoe_size']}<br>"
 
                      f"Grade: {row['grade']:.2f}<br>", axis=1)

for i, feature in enumerate(features):
    fig.add_trace(go.Scatter(
        x=df[feature],
        y=df["grade"],
        mode="markers",
        hovertext=hover_text,
        hoverinfo="text",
        name=feature,
        visible = (i == 0)
    ))

buttons = []
for i, feature in enumerate(features):
    visibility = [j == i for j in range(len(features))]
    buttons.append(dict(
        label=feature,
        method="update",
        args=[
            {"visible": visibility},
            {
                "title": {"text": f"Grade vs {feature}"},
                "xaxis": {"title": {"text": feature}}
            }
        ]
    ))
 
fig.update_layout(
        updatemenus=[dict(
        type="dropdown",
        direction="down",
        buttons=buttons,
        x=1.15,
        y=0.9
    )],
    title=f"Rating vs {features[0]}",
    xaxis_title=features[0],
    yaxis_title="Grade",
    showlegend=False
)

fig.show()
```
