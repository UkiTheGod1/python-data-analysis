## Interaktivni histogram pomocu Plotly-a

```python
import pandas as pd
import plotly.express as px
 
df = pd.read_csv("books.csv")
 
fig = px.histogram(df, x="rating", color="section", title="Distribution of Book Ratings")

fig.update_layout(xaxis_title="Rating", yaxis_title="Books (No.)", xaxis_range=[1, 5], bargap=0.1)
fig.show()
```

## Pregled i filtriranje - Lepši izgled
Šta se dešava samo kad dodamo "color" u fig?

```python
import pandas as pd
import plotly.express as px
 
df = pd.read_csv("books.csv")
 
fig = px.histogram(df, x="rating", color="section", title="Distribution of Book Ratings")

fig.update_layout(xaxis_title="Rating", yaxis_title="Books (No.)", xaxis_range=[1, 5], bargap=0.1)
fig.show()
```

## Refleksija
Dodavanjem samo "color" da odgovara atributu "section" dobijamo na različite boje ISTOG stuba histograma gde jasno vidimo koji procenat tog stuba su knjige iz određene sekcije. Sledeće nam je da dodamo box plot na postojeći histogram dodavajući atribut "marginal=box":

```python
import pandas as pd
import plotly.express as px
 
df = pd.read_csv("books.csv")
 
fig = px.histogram(df, x="rating", color="section", marginal="box", title="Distribution of Book Ratings")

fig.update_layout(xaxis_title="Rating", yaxis_title="Books (No.)", xaxis_range=[1, 5], bargap=0.1)
fig.show()
```

## Hover - Dodajemo još jaču inspekciju na ekstreme (hover data)

```python
import pandas as pd
import plotly.express as px
 
df = pd.read_csv("books.csv")
 
fig = px.histogram(df, x="rating", color="section", marginal="box",  hover_data=["title", "author", "ratings_count"], title="Distribution of Book Ratings")

fig.update_layout(xaxis_title="Rating", yaxis_title="Books (No.)", xaxis_range=[1, 5], bargap=0.1)
fig.show()
```
