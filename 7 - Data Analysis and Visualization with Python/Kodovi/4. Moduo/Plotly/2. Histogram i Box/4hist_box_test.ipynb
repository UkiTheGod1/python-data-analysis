Takođe, ne zaboravimo:

- **hover informacije** neka uključuju vrednost porudžbine i naziv proizvoda – da lako uočimo skrivene bisere među narudžbinama!

```python
import plotly.express as px
import pandas as pd

df = pd.read_csv("online_store_order_items.csv")

fig = px.histogram(df, x='order_value', color='category', marginal='box', hover_data=["name"], title='Order Values Distribution per Category')
fig.update_layout(xaxis_title="Order Value (€)", yaxis_title="Orders (No.)", bargap=0.1)
fig.show()
 
```
