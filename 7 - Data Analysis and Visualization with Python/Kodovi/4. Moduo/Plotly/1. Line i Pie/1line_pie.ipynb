## Uvod - Books - Number of books per section

```python
import plotly.express as px
import pandas as pd

df = pd.read_csv("books.csv")

section_counts = df["section"].value_counts().reset_index()
section_counts.columns = ["section", "count"]
 
fig = px.pie(data_frame=section_counts, names="section", values="count", title="Number of books per section")
fig.show()
```

## Raw - Number of order by delivery type - Bez CSV-a

```python
import plotly.express as px
import pandas as pd

delivery_types = ['Standard', 'Express', 'Pickup']
orders = [180, 70, 50]
 
df = pd.DataFrame({
    "delivery_type": delivery_types,
    "order_number": orders
})
 
fig = px.pie(data_frame=df, names="delivery_type", values="order_number", title="Number of orders by delivery type")
fig.show() 
```

## Average rating per century - Opsirnija analiza

```python
import pandas as pd
import plotly.express as px

df = pd.read_csv("books.csv")
df["published_century"] = ((df["year_published"] - 1) // 100 + 1).astype("Int64")
 
avg_rating_by_century = (
    df.groupby("published_century")["rating"].mean().reset_index()
)
 
fig = px.line(avg_rating_by_century,
              x="published_century",
              y="rating",
              markers=True,
              title="Average Book Rating by Century",
              labels={"published_century": "Century", "rating": "Average Rating"})

fig.update_traces(hovertemplate='Century: %{x}<br>Avg Rating: %{y:.2f}')
fig.show()
```
