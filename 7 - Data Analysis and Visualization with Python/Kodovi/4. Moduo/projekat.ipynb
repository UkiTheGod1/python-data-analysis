```python
import plotly.express as px
import pandas as pd
import plotly
plotly.offline.init_notebook_mode()

df = pd.read_csv("books.csv")
df.head(10)
```

## 🍰1. PieChart - Koliko knjiga ima svaka sekcija?

```python
section_counts = df['section'].value_counts().reset_index()
section_counts.columns = ['section', 'count']

fig = px.pie(section_counts, names="section", values="count", title="Number of Books per Section")
fig.show()
```

## 📈2. Linijski grafik - Ko se najviše čita?

```python
avg_borrowed_per_section = df.groupby("section")["times_borrowed"].mean().reset_index()
avg_borrowed_per_section.columns = ["section", "avg_times_borrowed"]
 
df_sorted = avg_borrowed_per_section.sort_values("avg_times_borrowed")

fig = px.line(df_sorted, x="section", y="avg_times_borrowed", markers=True, title="Average Number of Borrowings per Section")
 
fig.update_layout(xaxis_title="Section",
                  yaxis_title="Average Borrowings",
                  xaxis_tickangle=45)
fig.show()
```

## 📊4. Histogram + Boxplot - Ocene po sekcijama: Šta korisnici stvarno misle?

```python
fig = px.histogram(df, x="rating", color="section", marginal="box", hover_data=["title", "author", "ratings_count"], title="Rating per Section")
fig.update_layout(xaxis_title="Rating",
                  yaxis_title="Number of Books",
                  xaxis_range=[0, 5],
                  bargap=0.1)
fig.show()
```

## ⏺️ 5. Scatter Plot - Godina izdavanja vs. broj iznajmljivanja

```python
fig = px.scatter(df, x="year_published",
                     y="times_borrowed", 
                     hover_name="title", 
                     hover_data=["author"], 
                     title="Times borrowed per Year of Publishing",
                     trendline="lowess", # Teze vidljiva, manje bitna
                     color="section")    # Pomaze da se razlikuju sekcije - nije neophodno
fig.update_layout(xaxis_title="Year published",
                  yaxis_title="Times borrowed",
                  xaxis_range=[1400, 2025])
fig.show()
```
