# Zarada po mesecima detaljno po danima

```python
import pandas as pd
import plotly.express as px

df = pd.read_csv("D:/Python/libraries/prodaja_2025.csv")

df_group = df.groupby("Datum").agg(suma = ("Zarada", "sum")).reset_index().sort_values(by="Datum", ascending=True)

fig = px.line(df_group, x = "Datum", y = "suma", title="Zarada po mesecima", labels={"Datum": "Datum", "suma": "Zarada"}, template="plotly_dark")

fig.update_xaxes(
    dtick="M1",           # Prikazuj svaki mesec (M1 = 1 month)
    tickformat="%b",      # %b prikazuje samo skraćeno ime meseca (Jan, Feb...) - #B za pun naziv
    tickangle=0           # Drži nazive vodoravno
)

fig.update_traces(hovertemplate="<b>Datum:</b> %{x|%d. %b %Y}<br><b>Zarada</b>: %{y:,.0f}", line_color="#FFFC65")

fig.show()
```

# Zarada samo po mesecima ukupno

```python
import pandas as pd
import plotly.express as px

df = pd.read_csv("D:/Python/libraries/prodaja_2025.csv")

df["Datum"] = pd.to_datetime(df["Datum"], errors="coerce")
df["Mesec"] = df["Datum"].dt.month

df_group = df.groupby("Mesec").agg(suma = ("Zarada", "sum")).reset_index()

fig = px.line(df_group, 
              x = "Mesec", 
              y = "suma", 
              markers="o",
              title="Zarada po mesecima", 
              labels={"Mesec": "Mesec", "suma": "Zarada"}, 
              template="plotly_dark")

fig.update_traces(hovertemplate="<b>Mesec:</b> %{x}<br><b>Zarada</b>: %{y:,.0f}", line_color="#FFFC65")

fig.update_xaxes(nticks=13)
fig.update_layout(xaxis_range=[1.0, 12.0])

fig.show()
```
