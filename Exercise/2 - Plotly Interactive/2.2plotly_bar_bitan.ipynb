```python
import pandas as pd
import plotly.express as px

df = pd.read_csv("D:/Python/libraries/prodaja_2025.csv")
# 1. i 2. Grupisanje i sortiranje - Sortiranje smo obavili unutar grupacije jer uzmianje prvi 10 nije moguce zbog postojanja samo 5 proizvoda.

df_grouped = df.groupby("Proizvod").agg(suma = ("Zarada", "sum")).reset_index().sort_values(by="suma", ascending=False)

# 3. Crtanje grafika

fig = px.bar(df_grouped, 
             x = "suma", 
             y = "Proizvod", 
             color= "suma", 
             color_continuous_scale=["#20D600", "#FFFB26"], # Sluzi umesto palette (Magma, Viridis...), a mogu i custom hex
             title="Zarada po proizvodima",
             labels={"suma": "Zarada", "Proizvod": "Proizvod"},
             text_auto=".3s",
             template="plotly_dark")

fig.update_traces(hovertemplate=
                  "<b>Zarada</b>: %{x:,.0f}<br>"
                  "<b>Proizvod</b>: %{y}<br>"
                  "<extra></extra>")

# fig.update_yaxes(categoryorder="total ascending") - Barovi ce biti od najveceg ka namanjem
fig.update_layout(showlegend=False, coloraxis_showscale=False)  # Sklanja legendu vrednosti i boja
fig.show()


```
