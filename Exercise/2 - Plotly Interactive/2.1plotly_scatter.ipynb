```python
import plotly.express as px
import pandas as pd

df = pd.read_csv("D:/Python/libraries/prodaja_2025.csv")

df["Zarada"] = df["Cena_RSD"] * df["Količina"]

fig = px.scatter(df, 
                 x="Cena_RSD", 
                 y="Zarada", 
                 color="Grad",
                 title="Interaktivna analiza prodaje 2025",
                 hover_name="Proizvod",
                 hover_data=["Datum", "Grad"],
                 template="plotly_dark")

fig.update_traces(hovertemplate=
                "<b>Proizvod:</b> %{hovertext}<br>"
                "<b>Grad:</b> %{customdata[1]}<br>"
                "<b>Cena:</b> %{x:,.0f} RSD <br>"
                "<b>Zarada:</b> %{y:,.0f} RSD <br>"
                "<b>Datum:</b> %{customdata[0]} <br>"
                "<extra></extra>")

fig.show()
```
