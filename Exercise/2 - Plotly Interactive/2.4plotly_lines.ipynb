```python
import pandas as pd
import plotly.express as px

df = pd.read_csv("D:/Python/libraries/prodaja_2025.csv")

df["Datum"] = pd.to_datetime(df["Datum"], errors="coerce")

df_grouped = df.groupby(["Datum", "Proizvod"]).agg(zarada = ("Zarada", "sum")).reset_index()

fig = px.line(df_grouped, 
              x = "Datum", 
              y = "zarada", 
              color="Proizvod", 
              markers=True,
              hover_name="Proizvod", 
              title="Mesečna zarada po proizvodima", 
              labels={"Datum": "Datum", "zarada": "Zarada"},
              color_discrete_map={"Laptop": "#009DFF", "Miš": "#FBFF00", "Monitor": "#00FF04", "Slušalice": "#FF1414", "Tastatura": "#FF9D00"},
              template="plotly_dark")

fig.update_xaxes(
    dtick="M1",
    tickformat="%b",
    tickangle=0
)

fig.update_traces(hovertemplate="<b>Proizvod</b>: %{hovertext}<br>"
                "<b>Zarada</b>: %{y:,.0f}<br>"
                "<b>Datum</b>: %{x|%d. %B %Y}<br>"
                "<extra></extra>")

fig.update_yaxes(
    dtick=250000,            # Prikazuj podeoke na svakih x
    tickformat=".3s",        # Skraćeni format (1M, 2M...)
   # nticks=10,               # Alternativa: "Plotly, daj mi tačno 10 podeoka, ti izračunaj razmak"
   # gridcolor="gray",        # Boja horizontalnih linija
   # zerolinecolor="white"    # Istaknuta linija za nulu
)
fig.update_xaxes(
    range=["2025-01-01", "2025-12-31"], # Zaključaj osu na 2025. godinu
    nticks=12,                           # Sada će 12 podeoka pasti tačno na mesece
    tickformat="%b"                      # Samo ime meseca
)

fig.update_layout(hovermode="x unified")
fig.show()
```
