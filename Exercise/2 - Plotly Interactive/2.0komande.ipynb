| Komanda | Opis |

<h3 style="padding: 5px 15px; border-left: 5px solid #00eaff; background-color: #1a1a1a; color: #00eaff;">
    <u>fig.plot</u>
</h3>

| Komanda | Opis |
| :--- | :--- |
| **hover_name="xxx"** | Kod hoveringa, prikazuje ovu kolonu odvojeno i na vrhu |
| **hover_data=["xxx", "xxx"]** | Kod hoveringa, dodaje kolone koje nisu definisane sa x, y, color... |
| **opacity=x** | Definise providnost linija, stubica grafika - *od 0 do 1* |
| **template="plotly_dark"** | Menja boju pozadine grafika u crnu |

<h3 style="padding: 5px 15px; border-left: 5px solid #00eaff; background-color: #1a1a1a; color: #00eaff;">
    <u>fig.update_traces</u>
</h3>

| Komanda | Opis |
| :--- | :--- |
| **hovertemplate="XXX: %{hovertext}"** | Kod hovera, ***hovertext*** ce pristupiti koloni definisanoj u ***hover_name*** |
| **hovertemplate="XXX: %{fullData.name}** | Kod hovera, pristupa kategoriji kojom je bojen grafik |
| **hovertemplate="XXX: %{customdata[0]}"** | Kod hovera, ***customdata[n]*** ce pristupiti kolonama definisanim u ***hover_data*** |
| **hovertemplate=<extra<extra>></extra</extra>>** | Kod hovera, brise nepotreban ***side panel*** |
| **hovertemplate="XXX: %{x:.2f}"** | Zaokruzuje na 2 decimale |
| **hovertemplate="XXX: %{x:,.0f}"** | Ispisuje ceo broj odvojen zarezima (1,234,567) umesto skraceno (1.2M) |
| **hovertemplate="Date: %{x\|%d. %b %Y}"** | Kod hovera, formatira datum (npr. 26. Apr 2026) - *malo slovo je skraceni, veliko je puni format* |
| **marker_line_color="#XXXXXX"** | Podesava boju bordera |
| **marker_line_width=x** | Podesava debljinu bordera |

<h3 style="padding: 5px 15px; border-left: 5px solid #00eaff; background-color: #1a1a1a; color: #00eaff;">
    <u>fig.update_layout</u>
</h3>

| Komanda | Opis |
| :--- | :--- |
| **paper_bgcolor="#XXXXXX"** | Menja boju pozadine |
| **plot_bgcolor="#XXXXXX"** | Menja poju plota |
| **showlegend=False** | Brise legendu sa strane |
| **coloraxis_showscale=False** | Brise legendu gradacije boja sa strane |
| **xaxis_range=[x, y]** | Limitira x osu od odredjenog do odredjenog broja (isto i za y osu - yaxis_range) |
| **margin=dict(x=y)** | Postavlja marginu koja odvaja naslov od vrha prozora i od vrha grafika - *x = t,b,r,l (top, bottom, right, left), y=broj* |
| **title=dict(...)** | **Naslov:** <br> • `text`: tekst naslova <br> • `x/y`: pozicija (0.0 do 1.0) <br> • `xanchor`: "center", "left" ili "right" <br> • `font`: recnik (size, color) |
| **legend=dict(...)** | **Legenda:** <br> • `orientation`: "h" (horiz) ili "v" (vert) <br> • `x/y`: kordinate pozicije <br> • `xanchor/yanchor`: tačka sidrenja (center, left, top...) |

<h3 style="padding: 5px 15px; border-left: 5px solid #00eaff; background-color: #1a1a1a; color: #00eaff;">
    <u>fig.update_xaxes/fig.update_yaxes</u>
</h3>

| Komanda | Opis |
| :--- | :--- |
| **dtick="X"** | Prikazuje vrednosti na svakih X (npr. `dtick=100`) |
| **nticks=x** | Prikazuje tačno X vrednosti na osi (Plotly sam računa razmak) |
| **range=["x", "y"]** | Limitira vidljivi opseg (radi i za brojeve i za datume) |
| **tickformat=".Xs"** | Formatira brojeve (npr. `.2s` za 1.2M) |
| **tickangle=0** | Drži nazive vodoravno (sprečava rotaciju teksta) |

### 📅 Datumi i vreme
| Komanda | Opis |
| :--- | :--- |
| **dtick="M1"** | Prikazuje svaki mesec (M1=1 mesec, M3=kvartal) |
| **tickformat="%b"** | Skraćeno ime meseca (Jan, Feb). Koristi `%B` za pun naziv |

### 🏁 Grid (Mreža)
| Komanda | Opis |
| :--- | :--- |
| **showgrid=True** | Prikazuje linije mreže |
| **gridwidth=x** | Debljina linija mreže |
| **gridcolor="x"** | Boja mreže (npr. `"white"` ili `"rgba(255,255,255,0.1)"`) |
| **griddash="dash"** | Isprekidana linija mreže (kao `--`) |
| **zerolinecolor="x"** | Posebna boja za liniju koja označava nulu |

### ✒️ Fontovi i Nazivi
| Komanda | Opis |
| :--- | :--- |
| **title_text="x"** | Menja glavni naziv ose |
| **title_font=dict()** | Podešava `size` i `color` za **naziv** ose |
| **tickfont=dict()** | Podešava `size` i `color` za **vrednosti** (brojeve/tekst) na osi |
