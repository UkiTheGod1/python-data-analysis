```python
import pandas as pd
import plotly.express as px

df = pd.read_csv("D:/Python/libraries/prodaja_2025.csv")

df_sub = df[df["Proizvod"].isin(["Laptop", "Miš", "Monitor"])]

fig = px.histogram(df_sub, 
                   x = "Zarada", 
                   color="Proizvod", 
                   color_discrete_map={"Miš": "#E83131", "Laptop": "#EF901C", "Monitor": "#F1FF2C"},
                   hover_data=["Datum", "Cena_RSD", "Količina"],  # Za ekstreme (tackice) na boxplotu
                   # marginal="box", # Dobar je boxplot, ali kod naseg trenutnog eksperimentisanja sa x i y osom, ne ide
                   nbins=50)

fig.update_xaxes(title_text="Zarada (RSD)",
                title_font=dict(size=18, color='black'), # Naslov ose
                tickfont=dict(size=12, color='gray'),
                range=[0, 2600000], 
                dtick=100000, 
                tickangle=45, 
                tickformat=".2s")

fig.update_yaxes(title_text="Broj prodaja (Frekvencija)",
                title_font=dict(size=18, color='black'),
                dtick=25)

fig.update_layout(title=dict(
                        text="Distribucija zarade po proizvodima", 
                        x=0.5, 
                        xanchor="center", 
                        font=dict(size=24, color="#006426")),
                 bargap = 0.1,
                 legend=dict(
                        orientation="h",   # horizontalna legenda
                        yanchor="bottom",
                        y=1.0,            # iznad grafika, a ispod naslova
                        xanchor="center",
                        x=0.5
    ))

fig.show()
```
