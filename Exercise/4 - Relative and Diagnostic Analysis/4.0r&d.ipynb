<h3 style="padding: 5px 15px; border-left: 5px solid #00eaff; background-color: #1a1a1a; color: #00eaff;">
    <u>Merge</u>
</h3>

| Komanda | Opis |
| :--- | :--- |
| **pd.merge(df1, df2, left_on="x", right_on="y", how="z")** | `df1/df2`: dataframe koji spajamo <br> • `x/y`: zajednicke kolone po kojima spajamo <br> • `z`: "inner" - samo zajednicke, "left" - ostaju levi, "right" - ostaju desni, "outer" - ostaju svi |
| **df.drop(columns=["X"], inplace=True)** | Brise kolonu pod nazivom "X" jer nakon merge ostaju obe zajednicke kolone  |
